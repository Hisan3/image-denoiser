```python
#threshold
from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt


def calculate_features(image):
    hist, _ = np.histogram(
        image.ravel(),
        bins=256,
        range=(0, 256)
    )

    # Number of pixels at 0 and 255
    black_pixels = hist[0]
    white_pixels = hist[255]

    # Percentage of extreme pixels
    extreme_ratio = (
        black_pixels + white_pixels
    ) / image.size


    #Median Filter
    median = cv2.medianBlur(image, 3)

    # Difference between original and median image
    diff = cv2.absdiff(image, median)

    return hist, extreme_ratio, diff



# Extreme pixel ratio
extreme_thresholds = [
    0.01,
    0.02,
    0.05
]

# Median-filter changed pixel ratio
changed_thresholds = [
    0.005,
    0.01,
    0.02
]

# Difference intensity threshold
difference_thresholds = [
    50,
    80,
    120
]



for filename, image in images:

    hist, extreme_ratio, diff = calculate_features(image)

    print("IMAGE:", filename)

    print("Extreme pixel ratio:", extreme_ratio)


    print("\nDifference Threshold Test ")

    for d_threshold in difference_thresholds:

        changed_pixels = np.sum(diff > d_threshold)

        changed_ratio = (
            changed_pixels / image.size
        )

        print(
            f"Threshold = {d_threshold:3d} | "
            f"Changed ratio = {changed_ratio:.4f}"
        )


    # Test different changed-ratio thresholds
    print("\n Changed Ratio Threshold Test ")

    # Use 80 as the difference threshold
    changed_pixels = np.sum(diff > 80)

    changed_ratio = (
        changed_pixels / image.size
    )

    print("Changed ratio using difference threshold 80:",
          changed_ratio)

    for c_threshold in changed_thresholds:

        if changed_ratio > c_threshold:
            result = " Salt & Pepper"
        else:
            result = "Not salt & pepper"

        print(
            f"Threshold = {c_threshold:.3f} | "
            f"Result = {result}"
        )



  # Test different extreme-pixel thresholds
    print("\n Histogram Threshold Test")

    for e_threshold in extreme_thresholds:

        if extreme_ratio > e_threshold:
            result = "Possible Salt & Pepper"
        else:
            result = "Not Salt & Pepper"

        print(
            f"Threshold = {e_threshold:.3f} | "
            f"Result = {result}"
        )


# 5. GAUSSIAN NOISE MEASUREMENT
    gaussian_blur = cv2.GaussianBlur(
        image,
        (5, 5),
        0
    )

    residual = cv2.absdiff(
        image,
        gaussian_blur
    )

    residual_std = np.std(residual)

    print("\n Gaussian Threshold Test")

    gaussian_thresholds = [
        5,
        8,
        12
    ]

    print("Residual standard deviation:",
          residual_std)

    for g_threshold in gaussian_thresholds:

        if residual_std > g_threshold:
            result = "Possible Gaussian"
        else:
            result = "Possible No Noise"

        print(
            f"Threshold = {g_threshold:2d} | "
            f"Result = {result}"
        )


# 6. DISPLAY IMAGE + HISTOGRAM
    plt.figure(figsize=(12, 5))

    # Image
    plt.subplot(1, 2, 1)

    plt.imshow(
        image,
        cmap='gray'
    )

    plt.title("Input Image")
    plt.axis("off")


    # Histogram
    plt.subplot(1, 2, 2)

    plt.plot(hist)

    plt.title("Image Histogram")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Frequency")

    plt.xlim(0, 255)

    plt.show()
```

```python
from google.colab import files
import cv2
import numpy as np
import matplotlib.pyplot as plt


def detect_noise(image):
    # HISTOGRAM
    hist, _ = np.histogram(
        image.ravel(),
        bins=256,
        range=(0, 256)
    )

    black_pixels = hist[0]
    white_pixels = hist[255]

    extreme_ratio = (
        black_pixels + white_pixels
    ) / image.size


    # MEDIAN FILTER
    median = cv2.medianBlur(image, 3)

    difference = cv2.absdiff(
        image,
        median
    )

    # SALT & PEPPER MEASUREMENT
    changed_pixels = np.sum(
        difference > 80
    )

    changed_ratio = (
        changed_pixels / image.size
    )



    # GAUSSIAN NOISE MEASUREMENT
    gaussian_blur = cv2.GaussianBlur(
        image,
        (5, 5),
        0
    )

    residual = cv2.absdiff(
        image,
        gaussian_blur
    )

    residual_std = np.std(residual)


    # Salt & Pepper
    if changed_ratio > 0.01:

      noise = "Salt and Pepper Noise"

      reason = (
          "Many pixels change significantly after median filtering, indicating isolated black/white noise." )


    # Gaussian
    elif residual_std > 8:
        noise = "Gaussian Noise"
        reason = ("The image contains random intensity variations ")


    # No Noise
    else:
        noise = "No Noise"
        reason = ("There are few significant pixel changes after median filtering and the residual variation is low.")


    return (
        noise,
        reason,
        hist,
        extreme_ratio,
        changed_ratio,
        residual_std
    )



for filename, image in images:
    (
        noise,
        reason,
        hist,
        extreme_ratio,
        changed_ratio,
        residual_std
    ) = detect_noise(image)


    plt.figure(figsize=(12, 5))


    # Original image
    plt.subplot(1, 2, 1)

    plt.imshow(
        image,
        cmap='gray'
    )

    plt.title("Input Image")
    plt.axis("off")


    # Histogram
    plt.subplot(1, 2, 2)

    plt.plot(
        hist
    )

    plt.title("Image Histogram")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Frequency")
    plt.xlim(0, 255)


    plt.tight_layout()
    plt.show()


    # OUTPUT
    print("IMAGE:", filename)

    print(
        "Extreme Pixel Ratio:",
        round(extreme_ratio, 6)
    )

    print(
        "Changed Pixel Ratio:",
        round(changed_ratio, 6)
    )

    print(
        "Residual Standard Deviation:",
        round(residual_std, 4)
    )

    print(
        "Detected Noise:",
        noise
    )

    print(
        "Explanation:",
        reason
    )
```
